Identifying Similar Attributes in Publicly Available Datasets
To identify similar attributes between <private-geo-file>.xlsx, <private-order-file>.xlsx, train_df, and test_df, we need to analyze the structure and content of these datasets. 

Explanation
The function find_contextually_similar_columns compares the columns of two DataFrames and identifies columns that have similar names or contain similar keywords. It uses basic string matching to find columns that could be considered similar in context.

Key Points:
String Matching: The script uses simple substring matching to find columns that are contextually similar.
Flexibility: The function is designed to work with any two DataFrames, allowing you to compare different datasets easily.
Output: The output will give you a dictionary of columns that are contextually similar, which you can then use to further analyze or transform the data.

```python
import pandas as pd

# Load the provided datasets
address_with_geo = pd.read_excel(os.path.join(DATA_DIR, GEO_FILE))
sys_order_consolidation = pd.read_excel(os.path.join(DATA_DIR, SYS_ORDER_FILE), sheet_name=None)  # Load all sheets as a dictionary
train_df = pd.read_csv(os.path.join(DATA_DIR, 'train_df.csv'))
test_df = pd.read_csv(os.path.join(DATA_DIR, 'test_df.csv'))

# Display column names for comparison
print("Address with Geo Columns:", address_with_geo.columns)
print("Sys Order Consolidation Sheets and Columns:")
for sheet, data in sys_order_consolidation.items():
    print(f"  {sheet}: {data.columns}")
print("Train DF Columns:", train_df.columns)
print("Test DF Columns:", test_df.columns)

# Function to find contextually similar columns
def find_contextually_similar_columns(df1, df2):
    similar_columns = {}
    for col1 in df1.columns:
        for col2 in df2.columns:
            if col1.lower() in col2.lower() or col2.lower() in col1.lower():
                similar_columns[col1] = col2
    return similar_columns

# Compare train_df with address_with_geo
similar_geo_train = find_contextually_similar_columns(address_with_geo, train_df)
similar_geo_test = find_contextually_similar_columns(address_with_geo, test_df)
print("Contextually Similar Columns with Address with Geo in Train DF:", similar_geo_train)
print("Contextually Similar Columns with Address with Geo in Test DF:", similar_geo_test)

# Compare train_df with sys_order_consolidation
similar_sys_order_train = {}
similar_sys_order_test = {}
for sheet, data in sys_order_consolidation.items():
    similar_sys_order_train[sheet] = find_contextually_similar_columns(data, train_df)
    similar_sys_order_test[sheet] = find_contextually_similar_columns(data, test_df)
print("Contextually Similar Columns with Sys Order Consolidation in Train DF:", similar_sys_order_train)
print("Contextually Similar Columns with Sys Order Consolidation in Test DF:", similar_sys_order_test)
```

```python
import pandas as pd
from sklearn.cluster import DBSCAN
import numpy as np

# Load the datasets
train_df = pd.read_csv(os.path.join(DATA_DIR, 'train_df.csv'))
test_df = pd.read_csv(os.path.join(DATA_DIR, 'test_df.csv'))

# Convert relevant columns to datetime
train_df['Order_Date'] = pd.to_datetime(train_df['Order_Date'], dayfirst=True, errors='coerce')
train_df['Time_Order_picked'] = pd.to_datetime(train_df['Time_Order_picked'], format='%H:%M:%S', errors='coerce').dt.time
train_df['DateTime_Order_picked'] = pd.to_datetime(train_df['Order_Date'].astype(str) + ' ' + train_df['Time_Order_picked'].astype(str), errors='coerce')

test_df['Order_Date'] = pd.to_datetime(test_df['Order_Date'], dayfirst=True, errors='coerce')
test_df['Time_Order_picked'] = pd.to_datetime(test_df['Time_Order_picked'], format='%H:%M:%S', errors='coerce').dt.time
test_df['DateTime_Order_picked'] = pd.to_datetime(test_df['Order_Date'].astype(str) + ' ' + test_df['Time_Order_picked'].astype(str), errors='coerce')

# Compute total_full_pallets if not present
if 'total_full_pallets' not in train_df.columns:
    train_df['total_full_pallets'] = train_df['Time_taken(min)']  # Example alternative

if 'total_full_pallets' not in test_df.columns:
    test_df['total_full_pallets'] = test_df['Time_taken(min)']  # Example alternative

# Function to calculate distance between two points
def calculate_distance(lat1, lon1, lat2, lon2):
    return np.sqrt((lat1 - lat2)**2 + (lon1 - lon2)**2)

# Function to find consolidation opportunities using clustering
def find_consolidation_opportunities_optimized(df, eps=0.05, min_samples=2, time_window='1h'):
    locations = df[['Delivery_location_latitude', 'Delivery_location_longitude']].to_numpy()
    clustering = DBSCAN(eps=eps, min_samples=min_samples).fit(locations)
    df['cluster'] = clustering.labels_
    
    consolidation_opportunities = []
    
    for cluster_id in set(clustering.labels_):
        if cluster_id == -1:
            continue
        cluster_df = df[df['cluster'] == cluster_id].sort_values(by='DateTime_Order_picked')
        current_group = []
        for i in range(len(cluster_df)):
            current_order = cluster_df.iloc[i]
            if not current_group:
                current_group.append(current_order['ID'])
            else:
                previous_order = cluster_df.loc[cluster_df['ID'] == current_group[-1]].iloc[0]
                time_diff = (current_order['DateTime_Order_picked'] - previous_order['DateTime_Order_picked']).total_seconds() / 3600
                if time_diff <= pd.to_timedelta(time_window).total_seconds() / 3600:
                    combined_pallets = current_order['total_full_pallets'] + previous_order['total_full_pallets']
                    distance = calculate_distance(current_order['Delivery_location_latitude'], current_order['Delivery_location_longitude'],
                                                  previous_order['Delivery_location_latitude'], previous_order['Delivery_location_longitude'])
                    consolidation_opportunities.append({
                        'Order 1': current_order['ID'],
                        'Order 2': previous_order['ID'],
                        'Combined Pallets': combined_pallets,
                        'Distance (km)': distance
                    })
    return consolidation_opportunities

# Apply the optimized function to both train and test datasets
train_consolidation_opportunities_optimized = find_consolidation_opportunities_optimized(train_df)
test_consolidation_opportunities_optimized = find_consolidation_opportunities_optimized(test_df)

# Convert the lists of dictionaries to DataFrames
train_consolidation_opportunities_optimized_df = pd.DataFrame(train_consolidation_opportunities_optimized)
test_consolidation_opportunities_optimized_df = pd.DataFrame(test_consolidation_opportunities_optimized)

# Save the cleaned data to Excel files
train_consolidation_opportunities_optimized_df.to_excel(os.path.join(DATA_DIR, 'train_consolidation_opportunities_optimized.xlsx'), index=False)
test_consolidation_opportunities_optimized_df.to_excel(os.path.join(DATA_DIR, 'test_consolidation_opportunities_optimized.xlsx'), index=False)

# Provide file paths for download
train_file_path_opportunities_optimized = os.path.join(DATA_DIR, 'train_consolidation_opportunities_optimized.xlsx')
test_file_path_opportunities_optimized = os.path.join(DATA_DIR, 'test_consolidation_opportunities_optimized.xlsx')

(train_file_path_opportunities_optimized, test_file_path_opportunities_optimized)
```
